```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as mlt
import seaborn as sns
```

```python
df = pd.read_csv('/content/drive/MyDrive/transactions_for_fe.csv')
df
```

```python
df.dtypes
```

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'], errors = 'coerce')
df['SignupDate'] = pd.to_datetime(df['SignupDate'], errors = 'coerce')
df.dtypes
```

```python
print(f'There are {df.duplicated().sum()} duplicates')
print(f'Missing values in each column: {df.isnull().sum()}')
```

```python
# Calculate Total Sales on each order
df['TotalSales'] = df['Quantity'] * df['UnitPrice']
df
```

```python
# How many days (Tenure days) did it take customers to make their first purchase after they sign up?

df['TenureDays'] = df['OrderDate'] - df['SignupDate']
df
```

```python
# How many total orders does each customer have?
total_orders = df.groupby('CustomerID')['TransactionID'].count()
df['CustomerTotalOrder'] = df['CustomerID'].map(total_orders)

# Total amount each customer has ever spent, attached to every one of their rows
total_spent = df.groupby('CustomerID')['TotalSales'].sum()
df['CustomerLifetimeValue'] = df['CustomerID'].map(total_spent)

df.head(2)
```

```python
# #IsRepeatCustomer A trueor false flag: does this CustomerID have more than 1 order
def IsRepeatCustomer (total_order):
  if total_order > 1:
    return True
  else:
    return False


df['IsRepeatCustomer'] = df['CustomerTotalOrder'].apply(IsRepeatCustomer)
df
```

```python
# OrderMonth / OrderWeekday
df['OrderMonth'] = df['OrderDate'].dt.month_name()
df['OrderWeekday'] = df['OrderDate'].dt.day_name()
df.head()
```

**Analysis Question**

```python
#Do repeat customers spend more per order on average than one-time customers?
x = df.groupby('IsRepeatCustomer')[['TotalSales']].mean().round(2)
print(x)
```

```python
# Which acquisition Channel produces the most loyal (highest OrderCount) customers?
y = df.groupby('Channel')['TransactionID'].count()
print(y)
```
